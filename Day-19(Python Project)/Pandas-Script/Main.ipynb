### Task 1 — Read the Dataset
In this task, we will load the `ecommerce_sales.csv` file into a Pandas DataFrame.
Then, we will display the first few rows to understand the structure of the dataset.

```python
import pandas as pd 


df=pd.read_csv("Dataset/ecommerce_sales.csv")

print("Size :")
display(df.size)
print("First 5 Records:")
display(df.head(5))
print("Dataset Info:")
display(df.info)
print("Statistial info:")
display(df.describe())
print("Object type:")
display(df.dtypes)
```

### Task 2 — Inspect the Dataset
In this task, we will inspect the dataset to understand its structure, columns, and data types.
We will also check for missing values and duplicate records before starting the cleaning process.

```python
#check Null Values

print("Check for Null Values:")
display(df.isnull().sum())

print("Check for Duplicates:")
display(df.duplicated().sum())

#
```

### Task 3 — Clean the Dataset
In this task, we will clean the dataset by removing duplicate records and handling missing values appropriately.
After cleaning, we will verify that no unwanted duplicates or missing values remain.

```python
#Remove duplicates and null values  by adding new values..
#Since I have no duplicates and null values Check for any values that might cause trouble in the data set eg:age!=-negitve values
#if Any duplicates or null Values prresent use pd.fillna(value) 

print("IS Unit Price Negitive or not")
display(df[df['unit_price']<0]) #IF no records no Outliers

#Likewise Check for each Coloumn

# Check for invalid negative values

print("Negative Quantity:")
display(df[df["quantity"] < 0])

print("Negative Unit Price:")
display(df[df["unit_price"] < 0])

print("Negative Discount:")
display(df[df["discount"] < 0])

print("Negative Delivery Days:")
display(df[df["delivery_days"] < 0])

print("Negative Customer Rating:")
display(df[df["customer_rating"] < 0])

print("Negative Revenue:")
display(df[df["revenue"] < 0])

```

### Task 4 — Data Transformation
In this task, we will create new columns from the existing data to make the dataset more useful for business analysis.
We will calculate the total amount, discount amount, and final amount for each order.

```python

# 1.Calculate total amount before discount
df["total_amount"] = df["quantity"] * df["unit_price"]

# 2. Calculate discount amount
df["discount_amount"] = df["total_amount"] * df["discount"]

# 3. Calculate final amount after discount
df["final_amount"] = df["total_amount"] - df["discount_amount"]

# 4. Convert order_date to datetime
df["order_date"] = pd.to_datetime(df["order_date"])

# 5. Extract month from order date
df["order_month"] = df["order_date"].dt.month

# 6. Extract year from order date
df["order_year"] = df["order_date"].dt.year

# 7. Categorize delivery time
df["delivery_category"] = pd.cut(
    df["delivery_days"],
    bins=[0, 3, 7, float("inf")],
    labels=["Fast", "Normal", "Slow"]
)

# Display transformed data
display(
    df[
        [
            "order_id",
            "quantity",
            "unit_price",
            "discount",
            "total_amount",
            "discount_amount",
            "final_amount",
            "order_date",
            "order_month",
            "order_year",
            "delivery_category"
        ]
    ].head()
)
```

### Task 5 — Aggregate Data
In this task, we will calculate important statistical summaries such as total, average, minimum, maximum, and median values.
We will apply aggregation functions to understand overall sales, quantities, and customer ratings.

```python
# Quantity Summary
print("========== QUANTITY SUMMARY ==========")

print("Total Quantity Sold:", df["quantity"].sum())
print("Average Quantity per Order:", df["quantity"].mean())
print("Median Quantity per Order:", df["quantity"].median())
print("Minimum Quantity:", df["quantity"].min())
print("Maximum Quantity:", df["quantity"].max())
print("Number of Orders:", df["quantity"].count())


# Final Amount Summary
print("\n========== SALES SUMMARY ==========")

print("Total Sales:", df["final_amount"].sum())
print("Average Order Value:", df["final_amount"].mean())
print("Median Order Value:", df["final_amount"].median())
print("Minimum Order Value:", df["final_amount"].min())
print("Maximum Order Value:", df["final_amount"].max())
print("Number of Orders:", df["final_amount"].count())


# Customer Rating Summary
print("\n========== CUSTOMER RATING SUMMARY ==========")

print("Average Rating:", df["customer_rating"].mean())
print("Median Rating:", df["customer_rating"].median())
print("Minimum Rating:", df["customer_rating"].min())
print("Maximum Rating:", df["customer_rating"].max())
print("Number of Ratings:", df["customer_rating"].count())
```

### Task 6 — GroupBy Analysis
In this task, we will group the e-commerce data by product category, region, and payment method.
We will use aggregation functions to compare sales performance and order counts across different groups.

```python
# 1. Sales by Product Category
print("========== SALES BY PRODUCT CATEGORY ==========")

category_summary = df.groupby("product_category")["final_amount"].agg(
    ["sum", "mean", "count"]
)

display(category_summary)


# 2. Sales by Region
print("\n========== SALES BY REGION ==========")

region_summary = df.groupby("region")["final_amount"].agg(
    ["sum", "mean", "count"]
)

display(region_summary)


# 3. Sales by Payment Method
print("\n========== SALES BY PAYMENT METHOD ==========")

payment_summary = df.groupby("payment_method")["final_amount"].agg(
    ["sum", "count"]
)

display(payment_summary)
```

### Task 7 — Export the Results
In this task, we will save the cleaned and transformed dataset along with the summary results as CSV files.
This allows us to reuse the processed data and analysis results for future reporting or visualization.


```python
# Output location
output_path = "Result/"

# Save cleaned and transformed dataset
df.to_csv(
    output_path + "cleaned_ecommerce_sales.csv",
    index=False
)

# Save category summary
category_summary.to_csv(
    output_path + "category_summary.csv"
)

# Save region summary
region_summary.to_csv(
    output_path + "region_summary.csv"
)

# Save payment summary
payment_summary.to_csv(
    output_path + "payment_summary.csv"
)

print("All files exported successfully!")
print("Files saved in:", output_path)
```
