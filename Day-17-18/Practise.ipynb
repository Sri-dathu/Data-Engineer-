LOAD DATA FROM A DATASET

```python
#Import Necessary Libraries

import numpy as pnp
import pandas as pd

#Load a DataSet
studentdf= pd.read_csv("StudentPerformanceFactors.csv")
studentdf

```

```python
#Read the DataSet and Find about it..


studentdf.info() #Gives Entire Row
print("-------------------------------------------------------------------------------")
display(studentdf.describe()) #Statistical Data
print("-------------------------------------------------------------------------------")
#To Display first and Last Five Rows

print("First 5 Rows:")
display(studentdf.head(5))

print("Last 5 Rows:")
display(studentdf.tail(5))

display("Size:",studentdf.shape)
display("Columns",studentdf.columns)
display("Type",studentdf.dtypes)
```

```python
#Data Quality Check 

#Find missing Values
print("Null Values:")
display(studentdf.isnull().sum()) #To find missing values

print("Duplicate Rows:", studentdf.duplicated().sum())

#Find Attendence greater than 100 or not
print("Check for Attendance ")
display(studentdf['Attendance'].max())

print("Check for Negitive values:")
display(studentdf['Sleep_Hours'].min())
```

Data Cleaning and Transformation

```python

# 1. Remove Unwanted Columns

studentdf.drop(
    columns=[
        'Teacher_Quality',
        'Parental_Education_Level',
        'Distance_from_Home'
    ],
    inplace=True
)

print("Columns after removing unwanted columns:")
display(studentdf.head())


# 2. Handle Missing Values


print("Missing Values Before Handling:")
display(studentdf.isnull().sum())


# Fill numerical columns with median
numeric_columns = studentdf.select_dtypes(include='number').columns

for column in numeric_columns:
    studentdf[column] = studentdf[column].fillna(
        studentdf[column].median()
    )


# Fill categorical columns with mode
categorical_columns = studentdf.select_dtypes(include='object').columns

for column in categorical_columns:
    studentdf[column] = studentdf[column].fillna(
        studentdf[column].mode()[0]
    )


print("Missing Values After Handling:")
display(studentdf.isnull().sum())



# 3. Remove Duplicate Rows


print("Duplicate Rows Before Removing:",
      studentdf.duplicated().sum())

studentdf.drop_duplicates(inplace=True)

print("Duplicate Rows After Removing:",
      studentdf.duplicated().sum())



# 4. Check Data Types


print("Data Types:")
display(studentdf.dtypes)


# 5. Check Categorical Values


print("Gender Values:")
display(studentdf['Gender'].value_counts())

print("Parental Involvement Values:")
display(studentdf['Parental_Involvement'].value_counts())

print("Motivation Level Values:")
display(studentdf['Motivation_Level'].value_counts())

print("Family Income Values:")
display(studentdf['Family_Income'].value_counts())

print("School Type Values:")
display(studentdf['School_Type'].value_counts())



# 6. Check Invalid Values


print("Attendance Greater Than 100:")
display(studentdf[studentdf['Attendance'] > 100])

print("Negative Hours Studied:")
display(studentdf[studentdf['Hours_Studied'] < 0])

print("Negative Sleep Hours:")
display(studentdf[studentdf['Sleep_Hours'] < 0])

print("Invalid Previous Scores:")
display(
    studentdf[
        (studentdf['Previous_Scores'] < 0) |
        (studentdf['Previous_Scores'] > 100)
    ]
)

print("Invalid Exam Scores:")
display(
    studentdf[
        (studentdf['Exam_Score'] < 0) |
        (studentdf['Exam_Score'] > 100)
    ]
)



# 9. Final Data Validation


print("\n========== FINAL DATA VALIDATION ==========")

print("Dataset Shape:")
print(studentdf.shape)

print("\nMissing Values:")
display(studentdf.isnull().sum())

print("Total Missing Values:",
      studentdf.isnull().sum().sum())

print("\nDuplicate Rows:")
print(studentdf.duplicated().sum())

print("\nData Types:")
display(studentdf.dtypes)

print("\nFirst 10 Rows:")
display(studentdf.head(10))


 
                         
                         
```

```python
# Transform Creating new Features

studentdf['Performance_Category'] = pd.cut(
    studentdf['Exam_Score'],
    bins=[0, 40, 70, 100],
    labels=['Low', 'Medium', 'High']
)

studentdf.head(10)
```

```python
#Analysis 

Student_Performance = studentdf[
    [
        'Hours_Studied',
        'Attendance',
        'Sleep_Hours',
        'Previous_Scores',
        'Tutoring_Sessions',
        'Exam_Score',
        'Study_Category',
        'Performance_Category'
    ]
]

Student_Performance

# 1. SUM
print("Total Hours Studied:")
print(Student_Performance['Hours_Studied'].sum())

print("\nTotal Tutoring Sessions:")
print(Student_Performance['Tutoring_Sessions'].sum())



# 2. MEAN


print("\nAverage Hours Studied:")
print(Student_Performance['Hours_Studied'].mean())

print("\nAverage Attendance:")
print(Student_Performance['Attendance'].mean())

print("\nAverage Exam Score:")
print(Student_Performance['Exam_Score'].mean())


# 3. MEDIAN


print("\nMedian Hours Studied:")
print(Student_Performance['Hours_Studied'].median())

print("\nMedian Exam Score:")
print(Student_Performance['Exam_Score'].median())


# 4. MIN

print("\nMinimum Hours Studied:")
print(Student_Performance['Hours_Studied'].min())

print("\nMinimum Attendance:")
print(Student_Performance['Attendance'].min())

print("\nMinimum Exam Score:")
print(Student_Performance['Exam_Score'].min())


# 5. MAX

print("\nMaximum Hours Studied:")
print(Student_Performance['Hours_Studied'].max())

print("\nMaximum Attendance:")
print(Student_Performance['Attendance'].max())

print("\nMaximum Exam Score:")
print(Student_Performance['Exam_Score'].max())


# 6. COUNT


print("\nNumber of Exam Scores:")
print(Student_Performance['Exam_Score'].count())

print("\nNumber of Attendance Values:")
print(Student_Performance['Attendance'].count())

#7. SIZE


print("\nTotal Number of Rows:")
print(Student_Performance.size)



# 8. STANDARD DEVIATION


print("\nStandard Deviation of Exam Score:")
print(Student_Performance['Exam_Score'].std())

print("\nStandard Deviation of Attendance:")
print(Student_Performance['Attendance'].std())


# 9. VARIANCE


print("\nVariance of Exam Score:")
print(Student_Performance['Exam_Score'].var())

print("\nVariance of Attendance:")
print(Student_Performance['Attendance'].var())



# 10. UNIQUE


print("\nUnique Study Categories:")
print(Student_Performance['Study_Category'].unique())

print("\nUnique Performance Categories:")
print(Student_Performance['Performance_Category'].unique())



# 11. NUNIQUE


print("\nNumber of Study Categories:")
print(Student_Performance['Study_Category'].nunique())

print("\nNumber of Performance Categories:")
print(Student_Performance['Performance_Category'].nunique())



# 12. AGG


print("\nMultiple Aggregations on Exam Score:")

display(
    Student_Performance['Exam_Score'].agg(
        ['sum', 'mean', 'median', 'min', 'max', 'count', 'std', 'var']
    )
)



# 13. AGG ON MULTIPLE COLUMNS


print("\nMultiple Aggregations on Multiple Columns:")

display(
    Student_Performance.agg({
        'Hours_Studied': ['mean', 'min', 'max'],
        'Attendance': ['mean', 'min', 'max'],
        'Exam_Score': ['mean', 'median', 'min', 'max'],
        'Tutoring_Sessions': ['sum', 'mean', 'max']
    })
)
```

```python
#Group BY

Student_Performance.groupby('Study_Category')['Exam_Score'].mean()
```
